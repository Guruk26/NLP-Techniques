#### Remove html tags

```python
# Import regular expressions (re) module for regex-based pattern matching and substitution
import re
```

```python
# Sample text containing HTML markup tags (<h1>) to demonstrate HTML tag removal
text="<h1>Hello I am Gururaj<h1>"
```

```python
# Remove HTML tags using regex pattern <.*?>
# Logic: '<.*?>' matches any opening or closing angle bracket with any characters inside non-greedily
# and replaces them with an empty string
re.sub(r'<.*?>',"",text)
```

```python
# Create a sample DataFrame containing multiple rows with various HTML tags (<h1>, <br>, <p>)
data=pd.DataFrame({'col':['Hello <h1> I am Gururaj </h1>','<br>click here </br>','<p>How are you ? </p>']})
```

```python
# Display the sample DataFrame with raw HTML markup
data
```

```python
# Function to remove HTML tags across an iterable/series of text strings
# Logic: Iterates through each string in the input sequence, applies regex substitution, and returns cleaned list
def html_removal(text):
    empty_html=[]
    for i in text:
        empty_html.append(re.sub(r'<.*?>',"",i))
    return empty_html
```

```python
# Apply the html_removal function across columns of the DataFrame using .apply()
data.apply(html_removal)
```

#### Remove chats shortcuts

```python
# Lookup dictionary mapping informal chat abbreviations/slang to their full formal English phrases
slang_dict = {'asap':"As soon as possible",'gn':"good morning",'lol':"laugh out load"}
```

```python
# Sample DataFrame containing chat sentences with abbreviations ('asap', 'gn', 'lol')
pd.DataFrame({'slang':["Hey call me asap",'hey hello gn','lol that was funny']})
```

```python
# Sample test sentence containing slang 'asap'
text="Hey call me asap"
```

```python
# Function to expand slang words using dictionary lookup
# Logic: Splits sentence into words; looks up each word in slang_dict; replaces if found, keeps original if not
def expand_slang(text,slang_dict):
    words=text.split()
    return " ".join([slang_dict.get(word,word) for word in words])
```

```python
# Check total count of English stopwords in NLTK dataset (typically 179 words)
len(stopwords.words('english'))
```

```python
# Sample sentence containing a mix of stopwords ('how', 'are', 'you', 'i', 'am', 'was', 'for') and content words
text="Hey how are you i am fine I was likking for you"
```

```python
# Naive word tokenization using Python string split() on whitespace
text.split()
```

```python
# Manual stopword filtering logic:
# Iterates through tokens, replaces stopwords with empty string, keeps non-stopwords, copies list to 'x'
without_stops=[]
for i in text.split():
    if i in english:
        without_stops.append("")
    else:
        without_stops.append(i)
    x=without_stops[:]
    " ".join(x)
```

```python
# Inspect the resulting filtered token list 'x'
x
```

```python
# Function intended to remove stopwords from input text
# Logic: Checks word membership in text.split() and reconstructs space-separated string
def stopwords(text):
    li=[]
    if i in text.split():
        li.append("")
    else:
        li.append(i)
    x=li[:]
    li.clear()
    return " ".join(x)
```

#### Handling Emojis

```python
# Function to strip Unicode emojis using a compiled regular expression
# Logic: Defines character ranges covering emoticons, pictographs, transport symbols, and flag emojis
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
```

```python
# Create sample DataFrame containing conversational strings with various emojis
data = {
    'content': [
        "I love pizza 🍕 so much!",
        "Good morning 🌞 🌻",
        "Let's go for a ride 🚗 💨",
        "Vacation time 🏖️ ✈️",
        "Happy birthday 🎂 🎉",
        "Feeling strong 💪 🔥"
    ]
}

data=pd.DataFrame(data)
```

```python
# Function to convert emojis to textual descriptions (e.g., pizza emoji -> ':pizza:')
# Logic: Preserves semantic emotion in NLP by replacing graphical emojis with human-readable text tokens
def emoji_text(text):
    return emoji.demojize(text,delimiters=(" ", " "))
```
