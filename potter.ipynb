## Train a Wrod2Vec Model on our Personal Data

```python
# Section header: Corpus Data Ingestion & File Handling
# File Handling
```

```python
# Open and read Harry Potter Book 1 ('The Sorcerer's Stone') in read-text mode ('rt') into string 'h1'
with open('Book1.txt', 'rt') as b1:
    h1=b1.read()
```

```python
# Inspect raw text content of Book 1
h1
```

```python
# Open and read remaining Harry Potter Books 2 through 7 into variables h2 through h7
with open('Book2.txt', 'rt') as b2:
    h2=b2.read()
with open('Book3.txt', 'rt') as b3:
    h3=b3.read()
with open('Book4.txt', 'rt') as b4:
    h4=b4.read()
with open('Book5.txt', 'rt') as b5:
    h5=b5.read()
with open('Book6.txt', 'rt') as b6:
    h6=b6.read()
with open('Book7.txt', 'rt') as b7:
    h7=b7.read()
```

```python
# Inspect text content of Book 5 ('Order of the Phoenix')
h5
```

```python
# Concatenate text strings of all 7 books into a single consolidated corpus string 'data'
data=h1+h2+h3+h4+h5+h6+h7
```

```python
# Compute and display total character count across all 7 books (~6.4 million characters)
len(data) # Total charactors
```

```python
# Estimate total word count across all 7 books using whitespace split (~1.1 million words)
len(data.split()) # Total Words
```

```python
# Pipeline Overview:
# Applying NLP Preprocessing --> 1. Tokenization -> Sentinece Tokenization, 2. Stopword Removal, 3. Lowercasing and more basic preprocessing steps
```

```python
# Import NLTK sentence tokenizer and Gensim's simple_preprocess utility
# simple_preprocess lowercases, strips punctuation, and extracts word tokens between 2 and 15 characters
import nltk
from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
# Initialize empty list 'potter' to hold tokenized sentences
# Split full corpus string into individual sentences using NLTK's sent_tokenize()
potter=[]
data=sent_tokenize(data)
```

```python
# Display total count of sentences parsed from the 7 books (~80,000+ sentences)
len(data)
```

```python
# Iterate through each sentence string and apply gensim's simple_preprocess to tokenize words
# Each sentence is converted into a list of cleaned, lowercased word tokens
for i in data:
    potter.append(simple_preprocess(i))
```

```python
# Inspect preprocessed sentence token lists in 'potter'
potter
```

```python
# Section header: Stopword Filtering
# Stopword Removal
```

```python
# Import NLTK stopwords corpus
from nltk.corpus import stopwords
```

```python
# Load standard English stopwords into a Python set for fast O(1) membership lookup
stopwords=set(stopwords.words('english'))
```

```python
# Display the set of English stopwords
stopwords
```

```python
# Function to filter out English stopwords from a nested list of tokenized sentences
# Logic: Retains sentence boundaries (required by Word2Vec windowing) while removing non-informative words
def stopword_removal(text):
    cleaned_text=[]
    for sentence in text:
        new_sentence=[]
        for word in sentence:
            if word.lower() not in stopwords:
                new_sentence.append(word)
        cleaned_text.append(new_sentence)
    return cleaned_text
```

```python
# Execute stopword removal on all preprocessed sentences to produce 'cleaned_potter' corpus
cleaned_potter=stopword_removal(potter)
```

```python
# Section header: Custom Word2Vec Model Training
# Model Creation
```

```python
# Import gensim library for Word2Vec neural model training
import gensim
```

```python
# Instantiate custom Word2Vec model:
# window=10: Considers context of 10 words before and after target word (captures broad topical semantics)
# min_count=2: Discards words appearing less than 2 times to filter rare typos and low-frequency noise
model=gensim.models.Word2Vec(window=10,min_count=2)
```

```python
# Inspect initialized Word2Vec model object configuration
model
```

```python
# Build vocabulary index mapping and word frequency tables from cleaned Harry Potter sentences
# Note: Method call contains original typo 'biuld_vocab' (intended: build_vocab)
model.biuld_vocab(cleaned_potter)
```

```python
# Train the Word2Vec continuous vector weights across the Harry Potter corpus
# total_examples: total sentence count in corpus; epochs: number of training iterations over dataset
model.train(cleaned_potter,total_examples=model.corpus_count,epochs=model.epochs)
```

```python
# Look up learned embedding vector for main character 'harry' from KeyedVectors (.wv)
model.wv['harry']
```

```python
# Check dimensionality of custom trained word embeddings (Gensim default is 100 dimensions)
len(model.wv['harry'])
```

```python
# Query top 10 most semantically similar words to antagonist 'voldemort' based on narrative context
# Expected associations: 'riddle', 'death', 'dark', 'lord', etc.
model.wv.most_similar('voldemort')
```

```python
# Retrieve unit-normalized (L2 norm = 1.0) vector array for all vocabulary words
model.wv.get_normed_vectors()
```

```python
# Inspect shape of normalized vector matrix: (vocabulary_size, vector_dimensions)
model.wv.get_normed_vectors().shape
```

```python
# Extract full list of vocabulary words corresponding to the vector matrix row indices
y=model.wv.index_to_key
```

```python
# Import Principal Component Analysis (PCA) from scikit-learn for dimensionality reduction
from sklearn.decomposition import PCA
```

```python
# Initialize PCA to project high-dimensional embeddings (100-dim) down to 3 principal components
pca=PCA(n_components=3)
```

```python
# Fit PCA model on normalized word vectors and transform them into 3D coordinate points (X)
X=pca.fit_transform(model.wv.get_normed_vectors())
```
