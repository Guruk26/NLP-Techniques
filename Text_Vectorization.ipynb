```python
# Create a toy DataFrame of 4 short sentences to clearly trace CountVectorizer & TF-IDF matrix generation
d=pd.DataFrame({
    'col': [
        'People Love AI',
        'AI Love AI-ML',
        'AI-ML Love Love',
        'Love People People'
    ]
})
```

```python
# Display the toy corpus DataFrame
d
```

```python
# Import CountVectorizer from scikit-learn for Bag of Words (BoW) text representation
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Instantiate CountVectorizer with default unigram settings (converts text to word frequency counts)
cv=CountVectorizer()
```

```python
# Fit vectorizer on corpus and transform sentences into a sparse Document-Term Matrix (DTM)
cv.fit_transform(d['col'])
```

```python
# Convert sparse Document-Term Matrix to a dense NumPy 2D array
# Logic: Each row represents a document; each column represents the count of a specific vocabulary word
cv.fit_transform(d['col']).toarray()
```

```python
# View learned vocabulary dictionary mapping unique words to their column indices in the matrix
cv.vocabulary_
```

```python
# Instantiate CountVectorizer configured for Bigrams (ngram_range=(2,2))
# Logic: Extracts pairs of two consecutive words to capture local phrase context (e.g., 'people love', 'love ai')
cv=CountVectorizer(ngram_range=(2,2))
```

```python
# Transform corpus into a dense Document-Term Matrix containing frequency counts of bigrams
cv.fit_transform(d['col']).toarray()
```

```python
# View learned bigram vocabulary dictionary and corresponding column feature indices
cv.vocabulary_
```

```python
# Re-display toy corpus DataFrame for reference comparison against bigram matrix
d
```

## TF-IDF

```python
# Import TfidfVectorizer from scikit-learn for Term Frequency - Inverse Document Frequency representation
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Instantiate TfidfVectorizer
# Logic: Balances Term Frequency (TF: how often a word occurs in a document) with Inverse Document Frequency
# (IDF: how rare or unique the word is across the whole corpus), penalizing generic frequent words
tf=TfidfVectorizer()
```

```python
# Fit TfidfVectorizer on corpus and transform documents into a dense normalized TF-IDF score matrix
tf.fit_transform(d['col']).toarray()
```

```python
# Retrieve feature names (vocabulary words) sorted by column index
tf.get_feature_names_out()
```

```python
# View the learned Inverse Document Frequency (IDF) weights calculated for each vocabulary feature
# Logic: Words appearing across all documents get lower IDF scores; rare words get higher scores
tf.idf_
```
